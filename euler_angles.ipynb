```python
import numpy as np
import matplotlib.pyplot as plt
import mpl_toolkits.mplot3d
from ipywidgets import interact
import ipywidgets as widgets

# 오일러각(Roll, Pitch, Yaw)을 받아 3차원 회전 행렬을 계산하는 함수
def get_rotation_matrix(roll, pitch, yaw):
    # 각도를 라디안(Radian) 단위로 변환
    r = np.radians(roll)
    p = np.radians(pitch)
    y = np.radians(yaw)

    # Roll (X축 기준 회전 행렬)
    Rx = np.array([
        [1, 0, 0],
        [0, np.cos(r), -np.sin(r)],
        [0, np.sin(r), np.cos(r)]
    ])

    # Pitch (Y축 기준 회전 행렬)
    Ry = np.array([
        [np.cos(p), 0, np.sin(p)],
        [0, 1, 0],
        [-np.sin(p), 0, np.cos(p)]
    ])

    # Yaw (Z축 기준 회전 행렬)
    Rz = np.array([
        [np.cos(y), -np.sin(y), 0],
        [np.sin(y), np.cos(y), 0],
        [0, 0, 1]
    ])

    # 최종 회전 행렬 (Z-Y-X 순서: Rz * Ry * Rx)
    return Rz @ Ry @ Rx

# 3차원 공간에 좌표축의 회전을 그려주는 함수
def plot_euler(roll, pitch, yaw):
    fig = plt.figure(figsize=(7, 7))
    ax = fig.add_subplot(111, projection='3d')

    origin = np.array([0, 0, 0])

    # 회전하기 전 기본 단위 벡터 (X: 빨강, Y: 초록, Z: 파랑)
    axes_orig = np.eye(3)

    # 회전 행렬을 적용하여 새로운 축의 방향을 구함
    R = get_rotation_matrix(roll, pitch, yaw)
    axes_rot = R @ axes_orig

    # 화살표로 3차원 축 시각화
    # X축 (빨간색)
    ax.quiver(origin[0], origin[1], origin[2], axes_rot[0, 0], axes_rot[1, 0], axes_rot[2, 0], color='r', length=1.0, linewidth=2, label='X (Roll)')
    # Y축 (초록색)
    ax.quiver(origin[0], origin[1], origin[2], axes_rot[0, 1], axes_rot[1, 1], axes_rot[2, 1], color='g', length=1.0, linewidth=2, label='Y (Pitch)')
    # Z축 (파란색)
    ax.quiver(origin[0], origin[1], origin[2], axes_rot[0, 2], axes_rot[1, 2], axes_rot[2, 2], color='b', length=1.0, linewidth=2, label='Z (Yaw)')

    ax.set_xlim([-1.2, 1.2])
    ax.set_ylim([-1.2, 1.2])
    ax.set_zlim([-1.2, 1.2])
    ax.set_xlabel('X Axis')
    ax.set_ylabel('Y Axis')
    ax.set_zlabel('Z Axis')
    ax.set_title(f'Euler Angles\nRoll: {roll}°, Pitch: {pitch}°, Yaw: {yaw}°', fontsize=12)
    ax.legend()
    plt.show()

# 슬라이더로 조절할 수 있도록 연결
interact(plot_euler,
         roll=widgets.IntSlider(min=-180, max=180, step=5, value=0, description='Roll (X):'),
         pitch=widgets.IntSlider(min=-180, max=180, step=5, value=0, description='Pitch (Y):'),
         yaw=widgets.IntSlider(min=-180, max=180, step=5, value=0, description='Yaw (Z):'));
```
